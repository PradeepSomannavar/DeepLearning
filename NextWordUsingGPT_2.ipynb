```python
from transformers import pipeline, set_seed

generator = pipeline('text-generation', model='gpt2')
set_seed(42)

outputs = generator(
    "Happiness lies within and",
    max_length=30,          # 🔥 controls length (short)
    num_return_sequences=3,
    temperature=0.7,
    top_k=50,
    top_p=0.9,
    do_sample=True
)

for i, out in enumerate(outputs):
    text = out['generated_text']
    
    # take only first 2-3 lines
    short_text = "\n".join(text.split("\n")[:3])
    
    print(f"Output {i+1}:")
    print(short_text)
    print("-" * 40)
```

```python
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained('gpt2')  # load tokenizer

prompt = "Apollo 11 was the first crewed"
inputs = tokenizer(prompt, return_tensors='pt')  # tokenize

print(inputs)

# convert ids back to tokens
tokens = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])
print(tokens)
```

```python
from transformers import GPT2LMHeadModel

model = GPT2LMHeadModel.from_pretrained('gpt2')  # load model

outputs = model(**inputs)  # forward pass
outputs
```
